```python
# In [39]: Invers dan determinan matriks
a = np.array([(1, -1), (1, 1)])
print(a)
a_inv = np.linalg.inv(a)
print(a_inv)
det_a = np.linalg.det(a)
print(det_a)
```

```python
# In [40]: Solusi sistem persamaan linear
A = np.array([(1, -6), (2, -5)])
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)
# Cara 1: menggunakan invers
X1 = np.dot(A_inv, Y)
print(X1)

# Cara 2: menggunakan solve
X2 = np.linalg.solve(A, Y)
print(X2)
```

```python
# In [62]: Membuat grafik 3D Contour
def f(x, y):
    return np.cos(np.exp(-x**2 - y**2)) * np.sin(np.sqrt(x**2 + y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

```python
#Soal 2: Grafik 2 Dimensi (Kurva Hati / Heart Curve)
import matplotlib.pyplot as plt
import numpy as np

# Membuat grid koordinat x dan y
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Mendefinisikan persamaan
F = (
    X**6
    + 3 * X**4 * Y**2
    - 3 * X**4
    + 3 * X**2 * Y**4
    - X**2 * Y**3
    - 6 * X**2 * Y**2
    + 3 * X**2
    + Y**6
    - 3 * Y**4
    + 3 * Y**2
)

# Membuat plot grafik 2D
plt.figure(figsize=(7, 7))
plt.contour(X, Y, F, levels=[1], colors="red", linewidths=2.5)
plt.title(
    r"Grafik 2D: $x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1$",
    fontsize=10,
)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True, linestyle="--", alpha=0.6)
plt.gca().set_aspect("equal", adjustable="box")
plt.show()
```

```python
#Soal 3: Grafik 3 Dimensi
from mpl_toolkits import mplot3d
import matplotlib.pyplot as plt
import numpy as np

# Membuat grid x dan y
x = np.linspace(-1.0, 1.0, 300)
y = np.linspace(-0.5, 1.8, 300)
X, Y = np.meshgrid(x, y)

# Menghitung suku di dalam akar: 1 - x^2 - (y - sqrt(|x|))^2
inside_sqrt = 1.0 - X**2 - (Y - np.sqrt(np.abs(X))) ** 2

# Membatasi hanya area riil (di dalam kurva)
inside_sqrt[inside_sqrt < 0] = np.nan
Z_top = np.sqrt(inside_sqrt)
Z_bottom = -np.sqrt(inside_sqrt)

# Membuat visualisasi 3D
fig = plt.figure(figsize=(9, 7))
ax = plt.axes(projection="3d")

# Plot bagian atas dan bawah permukaan 3D
surf1 = ax.plot_surface(X, Y, Z_top, cmap="autumn", edgecolor="none", alpha=0.85)
surf2 = ax.plot_surface(
    X, Y, Z_bottom, cmap="autumn", edgecolor="none", alpha=0.85
)

ax.set_title(
    r"Grafik 3D: $x^2 + (y - \sqrt{|x|})^2 + z^2 = 1$", fontsize=12, pad=20
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.view_init(elev=25, azim=45)

plt.tight_layout()
plt.show()
```
